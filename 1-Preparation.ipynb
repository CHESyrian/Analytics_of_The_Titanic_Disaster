**Missing Values**

```python
def getMissingValues(df, perc=0):
    null_sum     = df.isna().sum().sort_values(ascending=False)
    rows_n       = df.shape[0]
    df_null_sum  = pd.DataFrame(
        {
            'Count'      : [null_sum[n] for n in null_sum.index],
            'Percentage' : [(null_sum[n] / rows_n)*100 for n in null_sum.index]
        },
        index=[n for n in null_sum.index]
    )
    df_null_sum = df_null_sum[df_null_sum['Percentage'] > perc]
    return df_null_sum
```

**`Age`**

```python
df_train.groupby(['Sex'])['Age'].mean()
```

```python
df_train.groupby(['Sex', 'Pclass'])['Age'].mean().unstack()
```

```python
fill_age_values = df_train.groupby(['Sex', 'Survived', 'Pclass'])['Age'].mean().unstack()
fill_age_values
```

```python
def fillNullAge(df):
    # Female & Unsurvived & Pclass 1
    df[(df['Sex']=='female') & (df['Survived']==0) & (df['Pclass']==1)] = \
        df[(df['Sex']=='female') & (df['Survived']==0) & (df['Pclass']==1)].fillna(value={'Age':np.ceil(fill_age_values.loc[('female', 0), 1])})
    
    # Female & Unsurvived & Pclass 2
    df[(df['Sex']=='female') & (df['Survived']==0) & (df['Pclass']==2)] = \
        df[(df['Sex']=='female') & (df['Survived']==0) & (df['Pclass']==2)].fillna(value={'Age':np.ceil(fill_age_values.loc[('female', 0), 2])})
    
    # Female & Unsurvived & Pclass 3
    df[(df['Sex']=='female') & (df['Survived']==0) & (df['Pclass']==3)] = \
        df[(df['Sex']=='female') & (df['Survived']==0) & (df['Pclass']==3)].fillna(value={'Age':np.ceil(fill_age_values.loc[('female', 0), 3])})
    
    # Female & Survived & Pclass 1
    df[(df['Sex']=='female') & (df['Survived']==1) & (df['Pclass']==1)] = \
        df[(df['Sex']=='female') & (df['Survived']==1) & (df['Pclass']==1)].fillna(value={'Age':np.ceil(fill_age_values.loc[('female', 1), 1])})
    
    # Female & Survived & Pclass 2
    df[(df['Sex']=='female') & (df['Survived']==1) & (df['Pclass']==2)] = \
        df[(df['Sex']=='female') & (df['Survived']==1) & (df['Pclass']==2)].fillna(value={'Age':np.ceil(fill_age_values.loc[('female', 1), 2])})
    
    # Female & Survived & Pclass 3
    df[(df['Sex']=='female') & (df['Survived']==1) & (df['Pclass']==3)] = \
        df[(df['Sex']=='female') & (df['Survived']==1) & (df['Pclass']==3)].fillna(value={'Age':np.ceil(fill_age_values.loc[('female', 1), 3])})
    
    # Male & Unsurvived & Pclass 1
    df[(df['Sex']=='male') & (df['Survived']==0) & (df['Pclass']==1)] = \
        df[(df['Sex']=='male') & (df['Survived']==0) & (df['Pclass']==1)].fillna(value={'Age':np.ceil(fill_age_values.loc[('male', 0), 1])})
    
    # Male & Unsurvived & Pclass 2
    df[(df['Sex']=='male') & (df['Survived']==0) & (df['Pclass']==2)] = \
        df[(df['Sex']=='male') & (df['Survived']==0) & (df['Pclass']==2)].fillna(value={'Age':np.ceil(fill_age_values.loc[('male', 0), 2])})
    
    # Male & Unsurvived & Pclass 3
    df[(df['Sex']=='male') & (df['Survived']==0) & (df['Pclass']==3)] = \
        df[(df['Sex']=='male') & (df['Survived']==0) & (df['Pclass']==3)].fillna(value={'Age':np.ceil(fill_age_values.loc[('male', 0), 3])})
    
    # Male & Survived & Pclass 1
    df[(df['Sex']=='male') & (df['Survived']==1) & (df['Pclass']==1)] = \
        df[(df['Sex']=='male') & (df['Survived']==1) & (df['Pclass']==1)].fillna(value={'Age':np.ceil(fill_age_values.loc[('male', 1), 1])})
    
    # Male & Unsurvived & Pclass 2
    df[(df['Sex']=='male') & (df['Survived']==1) & (df['Pclass']==2)] = \
        df[(df['Sex']=='male') & (df['Survived']==1) & (df['Pclass']==2)].fillna(value={'Age':np.ceil(fill_age_values.loc[('male', 1), 2])})
    
    # Male & Unsurvived & Pclass 3
    df[(df['Sex']=='male') & (df['Survived']==1) & (df['Pclass']==3)] = \
        df[(df['Sex']=='male') & (df['Survived']==1) & (df['Pclass']==3)].fillna(value={'Age':np.ceil(fill_age_values.loc[('male', 1), 3])})
    
    # Convert 'Age' to Integer
    df['Age'] = df['Age'].astype('int64')

    return df
```

**`Embarked`**

```python
def fillNanEmbarked(df):
    # Get Max Count
    max_embarked = df['Embarked'].value_counts().index[0]
    
    # Fill 'Embarked' with the Mode 'S'
    df['Embarked'] = df['Embarked'].fillna(max_embarked)

    return df
```

    Drop Outliers

```python
def getOutliersLimitis(data):
    Q1, Q3 = data.quantile([0.25, 0.75])
    IQR    = Q3 - Q1
    high   = Q3 + (1.5 * IQR)
    low    = Q1 - (1.5 * IQR)
    return (high, low)
```
