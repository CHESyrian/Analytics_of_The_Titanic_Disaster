#### **`Correlation`**

---

```python
corr = df_clean.select(cs.numeric()).corr()
```

```python
x = corr.columns
y = corr.columns
z = corr.to_numpy()

fig = go.Figure(
        data=go.Heatmap(
            x=x,
            y=y,
            z=z,
            colorscale='Viridis', 
            type='heatmap', 
            text=corr,
            texttemplate="%{text:.2}",
            textfont={"size":12}
        )
    )

fig.show()
```
